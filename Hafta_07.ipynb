```python
# SYSTEMATIC SAMPLE
import pandas as pd

df_coffee = pd.read_feather("resources/coffee_ratings_full.feather")

sample_size = 5
pop_size = len(df_coffee)
interval = pop_size//sample_size

shuffled = df_coffee.sample(frac=1).reset_index(drop=True)
systematic_sample = shuffled.iloc[interval::interval]

systematic_sample
```

```python
top_counts = df_coffee["country_of_origin"].value_counts()
top_counts
```

```python
top_counted_countries=["Mexico","Colombia","Guatemala","Brazil","Taiwan","United States (Hawaii)"]
coffee_ratings_top=df_coffee[df_coffee["country_of_origin"].isin(top_counted_countries)]
coffee_ratings_top
#sadece seçilen ülkelerdeki kahvelerdeki verilerimiz var.
```

```python
#basit rastgele örneklem 
# sample kullanarak veri setinin yüzde onluk basit rastgele örneklem seçimi
coffee_ratings_samp=coffee_ratings_top.sample(frac=0.1,random_state=42)
coffee_ratings_samp
```

```python
coffee_ratings_samp["country_of_origin"].value_counts(normalize=True)
```

```python
coffee_ratings_top.groupby("country_of_origin").count()
```

```python
coffee_rating_strat = coffee_ratings_top.groupby("country_of_origin").sample(frac=0.1,random_state=42)
coffee_rating_strat["country_of_origin"].value_counts(normalize=True)
```

```python
# STRATIFIED SAMPLE
total_samples = 20
df_size = len(coffee_rating_strat)

country_counts = coffee_rating_strat["country_of_origin"].value_counts()
country_props = country_counts / df_size

samples_per_country = (country_props * total_samples).astype(int)

stratified_sample = pd.DataFrame()

for country, count in samples_per_country.items():
    coffee_rating_eg = coffee_rating_strat[coffee_rating_strat["country_of_origin"] == country]
    selected = coffee_rating_eg.sample(n=count, replace=False, random_state=42)
    stratified_sample = pd.concat([stratified_sample, selected])
stratified_sample
```

```python
#tabakalı örneklemenin varyasyonu
#eşit oranlar yerine her gruptan eşit sayıda örnek alma
coffee_ratings_eg = coffee_ratings_top.groupby("country_of_origin").sample(n=15,random_state=2021)
coffee_ratings_eg["country_of_origin"].value_counts()
```

```python
# WEIGHTED RANDOM SAMPLE
import numpy as np
coffee_ratings_weight = coffee_ratings_top
#bu işlem ile taiwan olan satırların diğer satırlara göre iki kat seçilme şansı olacaktır
coffee_ratings_weight["weight"] = np.where(coffee_ratings_weight["country_of_origin"] == "Taiwan",2,1)
coffee_ratings_weight = coffee_ratings_weight.groupby("country_of_origin").sample(frac=0.1,weights="weight")
coffee_ratings_weight["country_of_origin"].value_counts(normalize=True)
```

```python
coffee_ratings_weight.head()
```

```python
# CLUSTER SAMPLING
import random
# kahve veri setindeki kahve çeşitlerini parametre olarak beklenir ve bazı grupları seçmek için basit rastgele alt örnekleme kullanılır.
varieties_samp = random.sample(list(df_coffee["variety"].unique()), k=3)

# sadece bu alt gruplarda basit rastgele alt örnekleme kullanılır.
coffee_ratings_cluster = df_coffee[df_coffee["variety"].isin(varieties_samp)]

# coffee_ratings_cluster veri çerçevesindeki "variety" sütununu kategorik değişkene dönüştürür.
# Yalnızca kullanılan kategorik değişkenleri tutar, kullanılmayanları kaldırır.
coffee_ratings_cluster.loc[:,"variety"] = coffee_ratings_cluster["variety"].astype("category").cat.remove_unused_categories()

# Örneklem oranlarının popülasyon oranları ile yakından eşleşmesi için örneklem alınmadan önce veriler kahve çeşitlerine göre gruplandırılır.
# Daha sonra beş adet yerine koyarak örnekleme(cluster sampling) yapılır.
coffee_ratings_cluster.groupby("variety", observed=True).sample(n=5, replace=True, random_state=42)["variety"]
```

```python
top_counted_countries = ['Mexico', 'Colombia','Guatemala', 'Brazil','Taiwan', 'United States (Hawaii)']

top_counted_subset = df_coffee['country_of_origin'].isin(top_counted_countries)

coffee_ratings_top = df_coffee[top_counted_subset]

coffee_ratings_top.shape
```

```python
coffee_ratings_srs= coffee_ratings_top.sample(frac=1/3,random_state=2021)
coffee_ratings_srs.shape
```

```python
coffee_rating_strat= coffee_ratings_top.groupby("country_of_origin").sample(frac=1/3,random_state=2021)
coffee_rating_strat.shape
```

```python
import random

top_countries_samp = random.sample(top_counted_countries, k=2)

top_condition = coffee_ratings_top['country_of_origin'].isin(top_countries_samp)

coffee_ratings_cls = coffee_ratings_top[top_condition]

coffee_ratings_cls.loc[:,'country_of_origin'] = coffee_ratings_cls['country_of_origin'].astype('category').cat.remove_unused_categories()

coffee_ratings_clst = coffee_ratings_cls.groupby('country_of_origin', observed=True).sample(n=len(coffee_ratings_top) // 6, replace=True)

coffee_ratings_clst.shape
```

```python
# relative errors
# popülasyonda parametresi ile nokta tahminleri arasındaki farkın mutlak değerine relative error denir.
```

```python
import numpy as np

df_error = pd.DataFrame(columns=['sample_size', 'relative_error'])

# Populasyon parametresi
population_mean = df_coffee['total_cup_points'].mean()

for num_of_rows in range(1, len(df_coffee) + 1):
    # Point Estimate
    sample_mean = df_coffee.sample(n=num_of_rows)['total_cup_points'].mean()
    relative_error = np.abs(population_mean - sample_mean) / population_mean
    df_error.loc[len(df_error.index)] = [num_of_rows, relative_error]

df_error.plot(x='sample_size', y='relative_error', kind='line')
```
