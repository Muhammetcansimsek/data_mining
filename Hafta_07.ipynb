```python
#basit rastgele örneklem 
# sample kullanarak veri setinin yüzde onluk basit rastgele örneklem seçimi
coffee_ratings_samp=coffee_ratings_top.sample(frac=0.1,random_state=2021)
coffee_ratings_samp
```

```python
coffee_rating_strat = coffee_ratings_top.groupby("country_of_origin").sample(frac=0.1,random_state=2021)
coffee_rating_strat["country_of_origin"].value_counts(normalize=True)
```

```python
country_counts = coffee_rating_strat["country_of_origin"].value_counts()
country_props = country_counts / len(coffee_rating_strat)
country_props
```

```python
total_samples = 20

samples_per_country = (country_props * total_samples).astype(int)

samples_per_country
```

```python
stratified_sample = pd.DataFrame()
for country, count in samples_per_country.items():
    coffee_rating_eg = coffee_rating_strat[coffee_rating_strat["country_of_origin"] == country]
    selected = coffee_rating_eg.sample(n=count, replace=False, random_state=42)
    stratified_sample = pd.concat([stratified_sample, selected], axis=0)
stratified_sample["country_of_origin"].value_counts()
```

```python
#relative errors
# popülasyonda parametresi ile nokta tahminleri ile arasındaki farka relative error denir. mutlak değerini alıyoruz
```

```python
import numpy as np

df_error = pd.DataFrame(columns=['sample_size', 'relative_error'])

pop_mean = df_coffee['total_cup_points'].mean()

for num_of_rows in range(1, len(df_coffee) + 1):
    rel_err = np.abs(pop_mean - df_coffee.sample(n=num_of_rows)['total_cup_points'].mean()) / pop_mean
    df_error.loc[len(df_error.index)] = [num_of_rows, rel_err]

df_error.plot(x='sample_size', y='relative_error', kind='line')
```
