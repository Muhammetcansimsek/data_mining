```python
#coffe rating dataset
import pandas as pd
df_coffee=pd.read_feather("resources/coffee_ratings_full.feather")
df_coffee
#tüm kahve türleri hakkında bilgi toplamışlar ve puan vermişler
```

```python
#df_coffee[df_coffee["total_cup_points"].max()]
#df_coffee.sort_values(by="total_cup_points", ascending=False).iloc[0]
df_coffee[df_coffee["total_cup_points"]==df_coffee["total_cup_points"].max()] #en iyi bu bence
```

```python
#pandas sapmple fınskiyonu satırların reasgele bir alt kümesini döndürür
pts_vs_flavor_pop=df_coffee[["total_cup_points","flavor"]]
pts_vs_flavor_pop
```

```python
#pandasın sample fonk. ile örneklem alırız.datafream ve series ile kullanılabilir
pts_vs_flavor_pop_samp = pts_vs_flavor_pop.sample(n=10)#replace true olmadığından aynı veriyi seçmez
#tek bir alan seçersek series döndürür,birden fazla alan seçersek dataframe döndürür.
pts_vs_flavor_pop_samp
```

```python
import numpy as np
print(f"popülasyon ortalaması ={np.mean(pts_vs_flavor_pop['total_cup_points'])}")
print(f"örnek ortalaması ={np.mean(pts_vs_flavor_pop_samp['total_cup_points'])}")
```
